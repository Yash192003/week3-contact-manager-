```python
import contacts_manager.ipynb as cm

def test_validate_phone():
    # valid number
    ok, cleaned = cm.validate_phone("9876543210")
    assert ok is True
    assert cleaned == "9876543210"
    # invalid number
    ok, cleaned = cm.validate_phone("abc123")
    assert ok is False
    assert cleaned is None

def test_validate_email():
    assert cm.validate_email("test@example.com") is True
    assert cm.validate_email("invalid-email") is False

def test_search_contacts():
    contacts = {
        "John Doe": {
            "phone": "1234567890",
            "email": "john@example.com",
            "address": "123 Main St",
            "group": "Friends",
            "created_at": "",
            "updated_at": ""
        },
        "Jane Smith": {
            "phone": "9999999999",
            "email": "jane@example.com",
            "address": "456 Oak Ave",
            "group": "Work",
            "created_at": "",
            "updated_at": ""
        },
    }
    results = cm.search_contacts(contacts, "john")
    assert "John Doe" in results
    assert "Jane Smith" not in results

if __name__ == "__main__":
    print("Run pytest for better output.")
```
